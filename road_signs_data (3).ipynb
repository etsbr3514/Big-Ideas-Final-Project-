```python
import pandas as pd
import numpy as np
from matplotlib.ticker import MaxNLocator

speedbumps = pd.read_csv("0010final_speedbumps.csv")
chart_speedbumps = speedbumps["neighborhood"].value_counts().plot.barh(xlabel="Speed Bumps",ylabel="Neighborhoods",)
chart_speedbumps.xaxis.set_major_locator(MaxNLocator(integer=True))

speedbumps["neighborhood"].value_counts()
```

```python
signs = pd.read_csv("0010final_signs.csv")
chart_signs = signs["neighborhood"].value_counts().iloc[::10].plot.barh(xlabel="Signs",ylabel="Neighborhoods",)
chart_signs.xaxis.set_major_locator(MaxNLocator(integer=True))

count = signs["neighborhood"].value_counts()
print(count.head(46))
print(count.tail(45))
```

```python
stop_signs = signs["description"] == "Stop"
yield_signs = signs["description"] == "Yield"
signs_slow = signs[stop_signs | yield_signs]

chart_signs_slow = signs_slow["neighborhood"].value_counts().iloc[::10].plot.barh(xlabel="Signs",ylabel="Neighborhoods",)
chart_signs_slow.xaxis.set_major_locator(MaxNLocator(integer=True))
signs_slow["neighborhood"].value_counts()
```
